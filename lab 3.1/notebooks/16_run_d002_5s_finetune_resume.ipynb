# Run D002 5s Fine-Tune

Bootstraps from `run_d002 epoch_006`, trains on ~5 second chunks (`max_frames=432`), saves resumable `latest.pt` state during epochs, and exports epoch samples from both cache-backed training clips and random Downloads songs.

```python
from __future__ import annotations

import json
import os
import subprocess
import sys
from datetime import datetime
from pathlib import Path

REPO = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
TAG = datetime.now().strftime('run_d002_5s_%Y%m%d_%H%M%S')
OUT_DIR = REPO / 'lab 3.1' / 'outputs' / 'diffusion_5s_finetune' / TAG
RUN_TRAIN = False

cmd = [
    sys.executable,
    str(REPO / 'lab 3.1' / 'scripts' / 'diffusion_longform_retool_train.py'),
    '--cache-dir', str(REPO / 'saves2' / 'lab3_diffusion' / 'run_d001' / 'cache'),
    '--out-dir', str(OUT_DIR),
    '--bootstrap-checkpoint', str(REPO / 'saves2' / 'lab3_diffusion' / 'run_d002' / 'checkpoints' / 'epoch_006.pt'),
    '--epochs', '6',
    '--batch-size', '1',
    '--grad-accum', '1',
    '--max-frames', '432',
    '--lr', '6e-5',
    '--identity-weight', '1.0',
    '--style-weight', '1.7',
    '--anchor-weight', '0.45',
    '--envelope-weight', '0.25',
    '--continuity-weight', '0.80',
    '--hf-penalty-weight', '0.24',
    '--vocal-weight', '0.55',
    '--crackle-weight', '0.38',
    '--anchor-bins', '40',
    '--hf-start-bin', '56',
    '--vocal-start-bin', '10',
    '--vocal-end-bin', '42',
    '--overlap-frames', '64',
    '--hf-margin', '0.05',
    '--crackle-margin', '0.010',
    '--style-probe-frames', '256',
    '--style-every-steps', '2',
    '--style-batch-splits', '1',
    '--max-batches-per-epoch', '4000',
    '--monitor-steps', '25',
    '--save-every-steps', '100',
    '--epoch-train-samples', '3',
    '--epoch-download-samples', '3',
    '--epoch-sample-ddim-steps', '50',
    '--epoch-sample-t-start', '260',
    '--epoch-sample-guidance-scale', '1.9',
    '--epoch-sample-style-strength', '0.62',
    '--resume',
    '--device', 'auto',
]

print(' '.join(map(str, cmd)))
print(f'Output dir: {OUT_DIR}')
```

```python
if RUN_TRAIN:
    env = os.environ.copy()
    env['PYTORCH_CUDA_ALLOC_CONF'] = 'expandable_segments:True'
    env['PYTHONIOENCODING'] = 'utf-8'
    subprocess.run(cmd, cwd=str(REPO), check=True, env=env)
else:
    print('Set RUN_TRAIN = True to launch the run.')
```
