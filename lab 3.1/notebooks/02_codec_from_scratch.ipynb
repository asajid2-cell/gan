# Codec Branch From Scratch

This notebook launches a fresh codec run from scratch rather than resuming from inherited checkpoints.

Use this notebook only after the audit notebook has been reviewed.

Notebook-first rules:
- use a fresh run name
- do not resume old runs
- export fixed-source samples every epoch
- run the realism supervisor at the end

```python
from pathlib import Path
import subprocess

REPO = Path.cwd().resolve().parent
RUN_NAME = 'run2001_codec_fresh'
SOURCE_FILE = r'C:\\Users\\Ahmed\\Downloads\\Milky & Mall Grab - Just The Way You Are.flac'
DO_RUN = False

cmd = [
    'python', str(REPO / 'lab 3' / 'run_lab3_codec.py'),
    '--run-name', RUN_NAME,
    '--force-custom-run-name',
    '--style-cond-source', 'mert_probe_embed',
    '--style-loss-mode', 'mert_probe_ce',
    '--translator-direct-output',
    '--translator-direct-mix', '0.35',
    '--stage1-epochs', '8',
    '--stage2-epochs', '12',
    '--stage3-epochs', '6',
    '--sample-count', '16',
    '--epoch-sample-source-file', SOURCE_FILE,
    '--epoch-sample-every', '1',
    '--epoch-sample-tag', 'epoch_samples',
]
cmd
```

```python
if DO_RUN:
    subprocess.run(cmd, check=True, cwd=REPO)
else:
    print('Set DO_RUN = True to launch the fresh codec training run.')
```
